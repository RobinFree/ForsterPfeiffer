```python
#Erstellt mit ChatGPT: 

import xml.etree.ElementTree as ET

def extract_places(xml_file):
    """Liest eine TEI-XML-Datei ein und extrahiert Ortsnamen."""
    tree = ET.parse(xml_file)
    root = tree.getroot()

    # TEI-Namespace (falls nötig)
    ns = {'tei': 'http://www.tei-c.org/ns/1.0'}

    # Suche nach allen <placeName>-Einträgen
    places = [place.text for place in root.findall(".//tei:placeName", ns) if place.text]

    return places

# Dateien einlesen
file1 = "pfeiffer_frauenfahrt01_1850.TEI-P5.xml"
file2 = "forster_reise01_1778.TEI-P5.xml"

places_pfeiffer = extract_places(file1)
places_forster = extract_places(file2)


print("\nOrte in Forsters Reise:")
for place in places_forster:
    print("-", place)

```

```python
# Import der relevanten Bibliotheken
from lxml import etree as et
import glob
import re
```

```python
# Sammeln der Ortsnamen in einer Liste, um die Liste dann in eine Tabelle umzuwandeln
rows = []


tree = et.parse('C:/Users/alina/OneDrive/Dokumente/Uni Basel/2. Semester/(De-)Coding History/Projekt/forster1.xml')
root = tree.getroot()

namespaces = {"tei": "http://www.tei-c.org/ns/1.0"}
   

places = root.findall(".//tei:placeName", namespaces=namespaces)

for place in places:
        row = []  # in dieser Liste sammeln wir alle Infos zum jeweiligen Ort

        # etwas viel Code, aber räumt uns schön die Ortsnennung auf, auch bei Zeilenumbrüchen etc.
        place_name = et.tostring(place, method="text", encoding="utf8", with_tail=False).decode("utf8")
        place_name = re.sub(r"\s*\n\s*", " ", place_name)
        row.append(place_name)

        rows.append(row)

print(rows[:3])
```

```python
from flair.nn import Classifier
from flair.data import Sentence


tagger = Classifier.load('ner')
```

```python

# Textdatei einlesen
with open("pfeiffer1_Text1.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Text in Sätze aufteilen
sentences_raw = [s.strip() for s in text.split(".") if s.strip()]

# Flair-Sentences erzeugen
sentences = [Sentence(s) for s in sentences_raw]

# In Batches taggen (beschleunigt Verarbeitung)
tagger.predict(sentences, mini_batch_size=16)

# Orte sammeln – in Reihenfolge, ohne Duplikate
orte = []
gesehen = set()

for sentence in sentences:
    for entity in sentence.get_spans('ner'):
        if entity.get_label("ner").value == "LOC":
            ort = entity.text
            if ort not in gesehen:
                orte.append(ort)
                gesehen.add(ort)

# Ausgabe in Reihenfolge
for ort in orte:
    print(ort)
```

Darstellung Pfeiffer mit Verbindungsstrichen um eine Route darzustellen: 

```python
import pandas as pd

# Einlesen der Tabelle
df = pd.read_csv('koordinaten_pfeiffer1.csv')

print(df.head)

import folium

# Tabelle Pfeiffer
df = df.rename(columns={'Column 1': 'place_name'})
df = df.dropna(subset=['Latitude', 'Longitude'])
df['Latitude'] = df['Latitude'].astype(float)
df['Longitude'] = df['Longitude'].astype(float)

# Initialisieren der Karte (Zentrum grob auf Deutschland)
m = folium.Map(location=[51.0, 10.0], zoom_start=5)

# Liste für die Koordinaten der Reiseroute
route_coords = []

# Marker setzen und Koordinaten für Linie sammeln
for _, row in df.iterrows():
    coord = [row['Latitude'], row['Longitude']]
    route_coords.append(coord)
    folium.Marker(
        location=coord,
        popup=row['place_name'],
        tooltip=row['place_name'],
        icon=folium.Icon(color="red", icon="info-sign")
    ).add_to(m)

# Linie zwischen den Punkten (Reiseroute)
folium.PolyLine(
    locations=route_coords,
    color='red',
    weight=3,
    opacity=0.8,
    tooltip='Reiseroute Pfeiffer'
).add_to(m)

# Karte anzeigen
m
```

Darstellung Pfeiffer mit Linien und einer Zeitachse: 

Für die Zeitachse wurden künstliche Zeitstempel verwendet. Sie zeigen lediglich in welcher Reihenfolge die Ortschaften erwähnt wurden und nicht wie lange der Aufenthalt oder die Wegstrecke war. 

```python
import pandas as pd

# Einlesen der Tabelle
df = pd.read_csv('koordinaten_pfeiffer1.csv')

print(df.head)

import folium
from folium.plugins import TimestampedGeoJson

# Tabelle Pfeiffer vorbereiten
df = df.rename(columns={'Column 1': 'place_name'})
df = df.dropna(subset=['Latitude', 'Longitude'])
df['Latitude'] = df['Latitude'].astype(float)
df['Longitude'] = df['Longitude'].astype(float)

# Künstliche Zeitstempel (nur Reihenfolge zählt)
df['timestamp'] = pd.date_range(start='2000-01-01', periods=len(df), freq='D')

# Initialisiere Karte
m = folium.Map(location=[51.0, 10.0], zoom_start=5)

# Statische Linie für komplette Route
route_coords = df[['Latitude', 'Longitude']].values.tolist()
route_coords = [[lat, lon] for lat, lon in route_coords]
folium.PolyLine(
    locations=route_coords,
    color='red',
    weight=3,
    opacity=0.8,
    tooltip='Reiseroute Pfeiffer'
).add_to(m)

# Zeitlich gesteuerte Punkte
features = []
for _, row in df.iterrows():
    feature = {
        'type': 'Feature',
        'geometry': {
            'type': 'Point',
            'coordinates': [row['Longitude'], row['Latitude']],
        },
        'properties': {
            'time': row['timestamp'].isoformat(),
            'popup': row['place_name'],
            'icon': 'circle',
            'iconstyle': {
                'fillColor': 'red',
                'fillOpacity': 0.8,
                'stroke': 'true',
                'radius': 6
            }
        }
    }
    features.append(feature)

# Zeitleisten-Plugin hinzufügen
TimestampedGeoJson({
    'type': 'FeatureCollection',
    'features': features,
}, 
period='P1D',
add_last_point=True,
auto_play=True,
loop=False,
max_speed=1,
loop_button=True,
date_options='YYYY-MM-DD',
time_slider_drag_update=True,
duration='P1D'
).add_to(m)

# Karte anzeigen
m
```

Darstellung Forster (blau) und Pfeiffer (rot) ohne Linien und mit Zeitachse: 

```python
import pandas as pd

# Einlesen der Tabelle
df = pd.read_csv('koordinaten_pfeiffer1.csv')

print(df.head)

import pandas as pd

# Einlesen der Tabelle
df2 = pd.read_csv('koordinaten_forster.csv')

print(df2.head)
import folium
from folium.plugins import TimestampedGeoJson
import datetime

# Tabelle Pfeiffer1 vorbereiten
df = df.rename(columns={'Column 1': 'place_name'})
df = df.dropna(subset=['Latitude', 'Longitude'])
df['Latitude'] = df['Latitude'].astype(float)
df['Longitude'] = df['Longitude'].astype(float)

# Füge einen künstlichen Zeitstempel hinzu (z. B. täglich beginnend ab heute)
start_time = datetime.datetime.now()
df['time'] = [start_time + datetime.timedelta(days=i) for i in range(len(df))]

# Tabelle Forster vorbereiten
df2 = pd.read_csv('koordinaten_forster.csv').rename(columns={
    'Column 1': 'place_name',
    'längengrad': 'Longitude',
    'breitengrad': 'Latitude'
})
df2 = df2.dropna(subset=['Latitude', 'Longitude'])
df2['Latitude'] = df2['Latitude'].astype(float)
df2['Longitude'] = df2['Longitude'].astype(float)
df2['time'] = [start_time + datetime.timedelta(days=i) for i in range(len(df2))]

# GeoJSON-Feature-Format vorbereiten
def df_to_geojson_features(dataframe, color):
    features = []
    for _, row in dataframe.iterrows():
        features.append({
            'type': 'Feature',
            'geometry': {
                'type': 'Point',
                'coordinates': [row['Longitude'], row['Latitude']],
            },
            'properties': {
                'time': row['time'].isoformat(),
                'popup': row['place_name'],
                'icon': 'circle',
                'iconstyle': {
                    'fillColor': color,
                    'fillOpacity': 0.8,
                    'stroke': 'true',
                    'radius': 6
                }
            }
        })
    return features

# GeoJSON-Daten erzeugen
features_all = df_to_geojson_features(df, 'red') + df_to_geojson_features(df2, 'blue')

# Karte initialisieren
m = folium.Map(location=[51.0, 10.0], zoom_start=4)

# Zeitstrahl hinzufügen
TimestampedGeoJson({
    'type': 'FeatureCollection',
    'features': features_all,
}, 
    period='P1D',  # 1 Tag Abstand
    add_last_point=True,
    auto_play=True,
    loop=False,
    max_speed=1,
    loop_button=True,
    date_options='YYYY-MM-DD',
    time_slider_drag_update=True
).add_to(m)

m
```
